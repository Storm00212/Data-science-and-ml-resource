```python
import torch
from torchvision import transforms
from PIL import Image
import matplotlib.pyplot as plt
```

# Defining the preprocessing pipeline

```python
transform = transforms.Compose([
    transforms.Resize((224, 224)),           # Resize image
    transforms.ToTensor(),                   # Convert to tensor (0–1 scale)
    transforms.Normalize(                    # Standardize
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])
```

```python
from google.colab import files
import io
from PIL import Image

print("Please upload your image file. For example, 'your_image.jpg'.")
uploaded = files.upload()

image_name = list(uploaded.keys())[0]
image = Image.open(io.BytesIO(uploaded[image_name])).convert("RGB") # Convert image to RGB

processed_image = transform(image)
```

```python
processed_image = processed_image.unsqueeze(0)
```

```python
print(processed_image.shape)
```

```python
from torchvision import models

model = models.resnet18(pretrained=True)
model.eval()

output = model(processed_image)

print(output.shape)
```
